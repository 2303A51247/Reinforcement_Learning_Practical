```python
import math

machines = ['A', 'B', 'C']

N_play = 10

N = {'A': 0, 'B': 0, 'C': 0}
R = {'A': 0, 'B': 0, 'C': 0}
Q = {'A': 0.0, 'B': 0.0, 'C': 0.0}
UCB = {'A': 0.0, 'B': 0.0, 'C': 0.0}

reward_patterns = {
    'A': [3, 4, 2, 5],
    'B': [6, 7, 5, 8],
    'C': [4, 3, 5, 4]
}

reward_index = {
    'A': 0,
    'B': 0,
    'C': 0
}

play_history = []

for play in range(1, N_play + 1):

    for machine in machines:

        if N[machine] == 0:
            UCB[machine] = float('inf')
        else:
            UCB[machine] = Q[machine] + math.sqrt(
                (2 * math.log(play)) / N[machine]
            )

    selected_machine = max(machines, key=lambda x: UCB[x])

    index = reward_index[selected_machine] % len(
        reward_patterns[selected_machine]
    )

    reward = reward_patterns[selected_machine][index]

    reward_index[selected_machine] += 1

    print("\nPlay", play)
    print("Selected Machine:", selected_machine)
    print("Reward:", reward)

    N[selected_machine] += 1

    R[selected_machine] += reward

    Q[selected_machine] = Q[selected_machine] + (
        (reward - Q[selected_machine]) / N[selected_machine]
    )

    play_history.append([
        play,
        selected_machine,
        reward,
        N['A'],
        N['B'],
        N['C'],
        Q['A'],
        Q['B'],
        Q['C']
    ])

for machine in machines:
    if N[machine] == 0:
        UCB[machine] = float('inf')
    else:
        UCB[machine] = Q[machine] + math.sqrt(
            (2 * math.log(N_play)) / N[machine]
        )

print("\n" + "=" * 70)
print("FINAL RESULTS")
print("=" * 70)

print("\nN TABLE – Participation Count")
print("--------------------------------")
for machine in machines:
    print(machine, ":", N[machine])

print("\nR TABLE – Total Reward")
print("--------------------------------")
for machine in machines:
    print(machine, ":", R[machine])

print("\nQ TABLE – Estimated Value")
print("--------------------------------")
for machine in machines:
    print(machine, ":", round(Q[machine], 3))

print("\nUCB TABLE")
print("--------------------------------")
for machine in machines:
    if UCB[machine] == float('inf'):
        print(machine, ": Infinity")
    else:
        print(machine, ":", round(UCB[machine], 3))

# Total reward received
total_reward = sum(R.values())

print("\nTotal Reward Received:", total_reward)

# Best machine
best_machine = max(machines, key=lambda x: Q[x])

print("Best Machine:", best_machine)

# Display complete play history
print("\n" + "=" * 70)
print("PLAY HISTORY")
print("=" * 70)

print(
    f"{'Play':<6}{'Machine':<10}{'Reward':<10}"
    f"{'N(A)':<8}{'N(B)':<8}{'N(C)':<8}"
    f"{'Q(A)':<10}{'Q(B)':<10}{'Q(C)':<10}"
)

print("-" * 70)

for row in play_history:
    print(
        f"{row[0]:<6}{row[1]:<10}{row[2]:<10}"
        f"{row[3]:<8}{row[4]:<8}{row[5]:<8}"
        f"{row[6]:<10.3f}{row[7]:<10.3f}{row[8]:<10.3f}"
    )
```
